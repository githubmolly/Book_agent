```python
import sys
from pathlib import Path

# Resolve project root (parent of notebooks/)
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))

from src.knowledge_base import KnowledgeBase

# Pass absolute path so it doesn't depend on where the notebook runs from
kb = KnowledgeBase(index_dir=str(PROJECT_ROOT / "data" / "index"))
print(f"Loaded {len(kb.chunks)} chunks from {PROJECT_ROOT / 'data' / 'index'}")

# Ask 5 questions where you know what the book covers
queries = [
    "What is retrieval augmented generation?",
    "How do you evaluate an LLM system?",
    "What is fine-tuning?",
    "What is AI as a judge?",
    "What are the steps of prompt engineering?"
]

for q in queries:
    print(f"\n{'='*60}\nQuery: {q}\n{'='*60}")
    results = kb.search(q, top_k=3)
    for r in results:
        print(f"\n[p.{r['page']} | score {r['score']:.2f}]")
        print(r["text"][:300] + "...")
```
